```python
import numpy as np
import matplotlib.pyplot as plt
import os
import cv2 # Required for connectedComponents

# ------------------------------- Function to color individual instances in a mask -------------------------------
def color_instances(mask):
    # Ensure mask is suitable for connectedComponents (binary or integer labels)
    # If the mask is already integer-labeled (e.g., from watershed), no conversion needed.
    # If it's a binary mask (0 or 1), connectedComponents will work to find distinct regions.
    processed_mask = mask.astype(np.uint8)

    num_labels, labels = cv2.connectedComponents(processed_mask)
    colored = np.zeros((mask.shape[0], mask.shape[1], 3), dtype=np.uint8)

    np.random.seed(42) # For reproducible colors

    for i in range(1, num_labels): # Skip background label (0)
        color = np.random.randint(0, 255, size=3)
        colored[labels == i] = color

    return colored


# Define the path to the .npy mask you want to view
# YOU CAN CHANGE THIS PATH TO ANY .npy MASK IN YOUR DRIVE
user_mask_path = '/content/drive/MyDrive/final_output_13_fixed/M_51_mask.npy'

if os.path.exists(user_mask_path):
    # Load the .npy mask
    mask = np.load(user_mask_path)

    # Apply coloring to the mask
    colored_mask = color_instances(mask)

    # Display the colored mask
    plt.figure(figsize=(5,5))
    plt.imshow(colored_mask)
    plt.title(f"Multi-colored NPY Mask: {os.path.basename(user_mask_path)}")
    plt.axis('off')
    plt.show()
else:
    print(f"Error: The specified mask file does not exist at {user_mask_path}")
```

```python
import os
import numpy as np
import cv2
from tensorflow.keras.utils import Sequence

class DataGenerator(Sequence):
    def __init__(self, img_dir, mask_dir, batch_size):
        self.img_dir = img_dir
        self.mask_dir = mask_dir
        self.batch_size = batch_size

        self.pairs = []

        # 🔥 Traverse B and M folders
        for category in ['B', 'M']:
            folder = os.path.join(img_dir, category)

            for file in os.listdir(folder):
                if not file.endswith(('.jpg', '.png', '.jpeg')):
                    continue

                num = os.path.splitext(file)[0]   # "12"

                if category == 'B':
                    mask_name = f"{num}_mask.npy"
                else:
                    mask_name = f"M_{num}_mask.npy"

                img_path = os.path.join(folder, file)
                mask_path = os.path.join(mask_dir, mask_name)

                if os.path.exists(mask_path):
                    self.pairs.append((img_path, mask_path))
                else:
                    print("❌ Missing mask:", mask_name)

        print("✅ Total valid pairs:", len(self.pairs))

    def __len__(self):
        return max(1, len(self.pairs) // self.batch_size)

    def __getitem__(self, idx):
        batch = self.pairs[idx*self.batch_size:(idx+1)*self.batch_size]

        images, masks = [], []

        for img_path, mask_path in batch:
            img = cv2.imread(img_path)
            mask = np.load(mask_path)

            if img is None:
                print("Skipping:", img_path)
                continue

            img = cv2.resize(img, (256, 256)) / 255.0

            mask = cv2.resize(mask, (256, 256),
                              interpolation=cv2.INTER_NEAREST)

            # ensure binary
            mask = (mask > 0).astype(np.uint8)

            images.append(img)
            masks.append(mask)

        if len(images) == 0:
            raise ValueError("❌ Empty batch!")

        return np.array(images), np.array(masks)
```

```python
def conv_block(x, f):
    x = layers.Conv2D(f, 3, padding='same', activation='relu')(x)
    x = layers.BatchNormalization()(x)
    x = layers.Conv2D(f, 3, padding='same', activation='relu')(x)
    x = layers.BatchNormalization()(x)
    return x

def attention_gate(x, g, f):
    x1 = layers.Conv2D(f, 1)(x)
    g1 = layers.Conv2D(f, 1)(g)

    add = layers.Add()([x1, g1])
    act = layers.Activation('relu')(add)

    psi = layers.Conv2D(1, 1, activation='sigmoid')(act)
    return layers.Multiply()([x, psi])

def attention_unet():
    inputs = layers.Input((IMG_SIZE, IMG_SIZE, 3))

    c1 = conv_block(inputs, 64); p1 = layers.MaxPooling2D()(c1)
    c2 = conv_block(p1, 128); p2 = layers.MaxPooling2D()(c2)
    c3 = conv_block(p2, 256); p3 = layers.MaxPooling2D()(c3)
    c4 = conv_block(p3, 512); p4 = layers.MaxPooling2D()(c4)

    c5 = conv_block(p4, 1024)

    u6 = layers.UpSampling2D()(c5)
    u6 = layers.Concatenate()([u6, attention_gate(c4, u6, 512)])
    c6 = conv_block(u6, 512)

    u7 = layers.UpSampling2D()(c6)
    u7 = layers.Concatenate()([u7, attention_gate(c3, u7, 256)])
    c7 = conv_block(u7, 256)

    u8 = layers.UpSampling2D()(c7)
    u8 = layers.Concatenate()([u8, attention_gate(c2, u8, 128)])
    c8 = conv_block(u8, 128)

    u9 = layers.UpSampling2D()(c8)
    u9 = layers.Concatenate()([u9, attention_gate(c1, u9, 64)])
    c9 = conv_block(u9, 64)

    outputs = layers.Conv2D(NUM_CLASSES, 1, activation='softmax')(c9)

    return Model(inputs, outputs)
```

```python
def convert_mask(mask):
    binary = (mask > 0).astype(np.uint8)

    kernel = np.ones((3,3), np.uint8)
    boundary = cv2.morphologyEx(binary, cv2.MORPH_GRADIENT, kernel)

    new_mask = np.zeros_like(mask)
    new_mask[binary == 1] = 1
    new_mask[boundary == 1] = 2

    return new_mask
```
